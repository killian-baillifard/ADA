# H0 : Gunshot deaths in the US

Reproduce the following claims made in the visualization :
- Nearly *two-thirds* of gun deaths are *suicides*.
- More than *85 percent* of suicide victims are *male*.
- Around *a third* of all gun deaths are *homicides*.
- Around *two-thirds* of homicide victims who are *males* in the *age-group of 15--34* are *black*.
- *Women* constitue only *15 percent* of the total *homicide* victims.

```python
import pandas as pd

df = pd.read_csv('./full_data.csv')
df.head()
```

```python
df['intent'].unique()
```

```python
num_entries = df.shape[0]
num_suicide = df[df['intent'] == 'Suicide'].shape[0]
print(f'Proportion of death by suicide in the US = {num_suicide / num_entries}')
```

```python
num_male_suicide_victims = df[(df['intent'] == 'Suicide') & (df['sex'] == 'M')].shape[0]
print(f'Proportion of male suicide in the US = {num_male_suicide_victims / num_suicide}')
```

```python
num_homicide = df[df['intent'] == 'Homicide'].shape[0]
print(f'Proportion of homicide in the US = {num_homicide / num_entries}')
```

```python
num_male_homicide_victims_15_34 = df[(df['intent'] == 'Homicide') & (df['sex'] == 'M') & (df['age'] >= 15) & (df['age'] <= 34)].shape[0]
num_black_male_homicide_victims_15_34 = df[(df['intent'] == 'Homicide') & (df['sex'] == 'M') & (df['age'] >= 15) & (df['age'] <= 34) & (df['race'] == 'Black')].shape[0]
print(f'Proportion of homicide victims who are males in the age-group of 15-34 and black = {num_black_male_homicide_victims_15_34 / num_male_homicide_victims_15_34}')
```

```python
num_female_homicide = df[(df['intent'] == 'Homicide') & (df['sex'] == 'F')].shape[0]
print(f'Proportion of female homicide victims : {num_female_homicide / num_homicide}')
```
